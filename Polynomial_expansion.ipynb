# Polynomial Expansion
Expande los polinomios resultantes del analisis de estabilidad usando la condicion de Routh-Hurwitz (Seccion 3.3 del paper) para analizar si los terminos son positivos o no.

```python
import sympy as sp
```

# Funciones

```python

def check(expr, vars_, name):
# Imprime cuántos términos tiene `expr` y si alguno tiene coeficiente negativo.
    P = sp.Poly(expr, *vars_)
    neg = [(mon, c) for mon, c in P.terms() if c < 0]
    print(f"{name}: {len(P.terms())} términos, {len(neg)} con coeficiente negativo")
    if neg:
        print("  negativos:", neg)
    else:
        print("  todos los coeficientes son positivos -> expresión > 0")

def hurwitz_exprs(Ae, B, C, m, a0):
# Devuelve (a3*a2 - a1, a3*a2*a1 - a1^2 - a3^2*a0) expandidas.
    a3 = Ae + B + C + m
    a2 = m*(Ae + B + C) + Ae*B + Ae*C + B*C
    a1 = m*(Ae*B + Ae*C + B*C) + Ae*B*C
    return (sp.expand(a3*a2 - a1),
            sp.expand(a3*a2*a1 - a1**2 - a3**2*a0))
```

# Condiciones

```python
N1 = "a3*a2 - a1"
N2 = "a3*a2*a1 - a1^2 - a3^2*a0"
```

# EP1

```python
# Caso Q0 = 1 - r  (0 < r < 1)
# A = σe'+μe, B = σl+μl, C = σp+μp
A, B, C, m, r = sp.symbols("A B C mu_m r", positive=True)

expr1, expr2 = hurwitz_exprs(A, B, C, m, a0=m*A*B*C*r)   # a0 = μm·A·B·C·(1 - Q0)

# Expresión 1: no depende de r
check(expr1, (A, B, C, m), N1)

# Expresión 2: r solo aparece en -a3^2*a0, así que es decreciente en r
# y el peor caso es r = 1 (Q0 = 0).
check(sp.expand(expr2.subs(r, 1)), (A, B, C, m), f"{N2}  (peor caso r=1)")
```

# EP2

```python
# Caso Q0 = 1 + q  (q > 0)
A, B, C, m, q = sp.symbols("A B C mu_m q", positive=True)

Ae = A*(1 + q)   # σe'+μe+½φM/K = A·Q0
expr1, expr2 = hurwitz_exprs(Ae, B, C, m, a0=m*A*B*C*q)  # a0 = μm·A·B·C·(Q0-1)

check(expr1, (A, B, C, m, q), N1)
check(expr2, (A, B, C, m, q), N2)
```
